```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Generate a synthetic dataset
np.random.seed(0) # For reproducibility
x = np.random.rand(100, 1) # 100 samples
y = 3 * x + 5 + np.random.randn(100, 1) # y = 3x + 5 + noise
```

```python
# Visualize the synthetic dataset with a scatter plot
plt.scatter(x, y)
plt.title('Synthetic Dataset')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f'Training data size: {X_train.shape[0]} samples')
print(f'Testing data size: {X_test.shape[0]} samples')
```

```python
# Train a linear regression model on the training data
model = LinearRegression()
model.fit(X_train, y_train)

# Display the coefficients
print(f'Coefficient: {model.coef_}')
print(f'Intercept: {model.intercept_}')
```

```python
# Evaluate the model on the test data
y_pred = model.predict(X_test)

# Calculate and print the R² score
r2 = r2_score(y_test, y_pred)
print(f'R² score: {r2}')

# Calculate and print the mean squared error
mse = mean_squared_error(y_test, y_pred)
print(f'Mean Squared Error: {mse}')
```

```python
# Visualize the original data points along with the regression line
plt.scatter(X_test, y_test, color='blue', label='Actual data')
plt.plot(X_test, y_pred, color='red', label='Regression line')
plt.title('Linear Regression Result')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
